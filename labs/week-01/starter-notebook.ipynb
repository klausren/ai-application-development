## Part 2 · Your First ML Program — Iris (15 min)

The Iris dataset: 150 flowers, 4 measurements (sepal/petal length & width), 3 species. The "Hello World" of machine learning.

The scikit-learn workflow you saw in the lecture:

```
fit  →  predict  →  evaluate
```

Run the cells, read the comments, and watch a machine learn in under a second.


````python
# Cell 2.1 - load and look at the data. ALWAYS look before you model.
from sklearn.datasets import load_iris

iris = load_iris()
X, y = iris.data, iris.target

print("Data shape  :", X.shape, "  (150 samples x 4 features)")
print("Feature names:", iris.feature_names)
print("Classes      :", iris.target_names)
print()
print("First 3 rows of X:\n", X[:3])
print("First 3 labels:", y[:3])
````

````python
# Cell 2.2 - split: train on 70%, test on 30%.
# random_state=42 makes the split reproducible — everyone gets the same split.
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=42, stratify=y
)
print("Train:", X_train.shape, "  Test:", X_test.shape)
````

````python
# Cell 2.3 - Model #1: Logistic Regression. fit -> predict -> evaluate.
from sklearn.linear_model import LogisticRegression

logreg = LogisticRegression(max_iter=200)
logreg.fit(X_train, y_train)                 # <- this single line is "training"

acc_logreg = logreg.score(X_test, y_test)    # accuracy on unseen test data
print(f"LogisticRegression test accuracy: {acc_logreg:.4f}")
````
